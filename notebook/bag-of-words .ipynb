````python
# 将一行生词转换为对应词的字符串
# input：一行生词
# output：对应词的字符串
def review_to_words(raw_review):
    #1. Remove HTML
    review_text = BeautifulSoup(raw_review).get_text()

    #2. Remove non-letters
    letters_only = re.sub("[^a-zA-Z]"," ",review_text)

    #3. Convert to lower case, split into individual words
    words = letters_only.lower().split()

    #4. In python, searching a set is much faster than searching
    stops = set(stopwords.words("english"))

    #5. Remove stop words
    meaningful_words=[w for w in words if not w in stops]

    #6. Join the words back into one string separated by space
    return(" ".join(meaningful_words))
````

## 从词袋创建特征（使用 scikit-learn）

现在我们已经把训练评论整理好了，如何将它们转换为某种数值表示形式以供机器学习使用呢？一种常见的方法叫做**词袋模型（Bag of Words）**。

词袋模型从所有文档中学习一个词汇表，然后通过统计每个单词在每篇文档中出现的次数来对文档进行建模。例如，考虑以下两个句子：

**句子1：** `"The cat sat on the hat"`

**句子2：** `"The dog ate the cat and the hat"`

从这两个句子中，我们得到的词汇表如下：

`{ the, cat, sat, on, hat, dog, ate, and }`

为了得到词袋表示，我们统计每个单词在每个句子中出现的次数。在句子1中，`"the"` 出现了两次，`"cat"`、`"sat"`、`"on"` 和 `"hat"` 各出现一次，因此句子1的特征向量为：

| | the | cat | sat | on | hat | dog | ate | and |
|---|---|---|---|---|---|---|---|---|
| **句子1** | 2 | 1 | 1 | 1 | 1 | 0 | 0 | 0 |

同样地，句子2的特征向量为：`{ 3, 1, 0, 0, 1, 1, 1, 1 }`

在 IMDB 数据中，我们有非常多的评论，这将产生一个非常大的词汇表。为了限制特征向量的大小，我们应该选择一个最大的词汇量。下面，我们使用**5000个最频繁出现的词**（记住，停用词已经被移除了）。

我们将使用 scikit-learn 的 `feature_extraction` 模块来创建词袋特征。。

## 词袋模型示意图

```
评论1: "I loved this movie"     →  [1, 1, 1, 1, 0, 0, ...]
评论2: "This movie was great"   →  [0, 1, 1, 0, 1, 1, ...]
评论3: "Great film"             →  [0, 0, 0, 0, 1, 0, ...]
        ↑   ↑   ↑   ↑   ↑
       i  loved this movie great ...
```

每一行是一篇评论，每一列是一个单词，值是该词在该评论中出现的次数。

| **特征向量** | 每个文档对应一个向量，长度为词汇表大小 |

剩下的工作就是在我们的测试集上运行训练好的随机森林模型，并创建提交文件。`testData.tsv`这个文件包含了另外 25,000 条评论及其对应的 id；我们的任务是预测它们的情感标签。

**请注意**，当我们在测试集上使用词袋模型时，我们只调用 `transform` 方法，而不是像训练集那样调用 `fit_transform`。在机器学习中，你不应该用测试集来拟合（训练）你的模型，否则会有过拟合的风险。因此，在准备好进行预测之前，我们不会触碰测试集。

**`transform` vs `fit_transform` 的核心原因**
    *   **训练集**：使用 `fit_transform`。它做了两件事：① 学习并构建词汇表（`fit`）；② 将文本转换为特征向量（`transform`）。
    *   **测试集**：只使用 `transform`。它使用训练集上已经构建好的**同一个词汇表**来转换数据。这保证了训练和测试数据具有相同的特征维度，且没有数据泄露。

## 引入分布式词向量

本教程的这一部分将重点介绍如何使用由 Word2Vec 算法创建的分布式词向量。

Word2Vec 由 Google 于 2013 年发布，是一种神经网络实现，用于学习词的分布式表示。在此之前，也有其他深度或循环神经网络架构被提出用于学习词表示，但主要问题是训练模型所需的时间太长。Word2Vec 相对于其他模型学习速度更快。

Word2Vec 不需要标签就能创建有意义的表示。这非常有用，因为现实世界中的大多数数据都是无标签的。如果给网络足够多的训练数据（数百亿个词），它就能产生具有有趣特性的词向量。含义相似的词会聚集在一起，而聚类之间的距离使得一些词之间的关系（例如类比关系）可以通过向量数学运算来重现。最著名的例子是，使用训练良好的词向量，可以实现 `"king - man + woman = queen"`。

分布式词向量功能强大，可用于许多应用，尤其是词预测和翻译。在这里，我们将尝试将它们应用于情感分析。

## 为什么 Word2Vec 要保留停用词？

```
原句: "The movie was not good at all"

移除停用词后: ["movie", "not", "good"]
保留停用词后: ["the", "movie", "was", "not", "good", "at", "all"]
```

Word2Vec 通过词与词的**共现关系**（上下文窗口）来学习词向量。停用词作为上下文的一部分，有助于模型理解词的分布规律。例如，`"not good"` 中的 `"not"` 对情感判断至关重要。

````python
import pandas as pd

train = pd.read_csv("/kaggle/input/datasets/kintsugi0v0/imdb-data/labeledTrainData.tsv", header=0, delimiter="\t", quoting=3)
test = pd.read_csv("/kaggle/input/datasets/kintsugi0v0/imdb-data/testData.tsv", header=0, delimiter="\t", quoting=3)
unlabeled_train = pd.read_csv("/kaggle/input/datasets/kintsugi0v0/imdb-data/unlabeledTrainData.tsv", header=0, delimiter="\t", quoting=3)

print("Read %d labeled train reviews, %d labeled test reviews and %d unlabeled reviews\n" % (train["review"].size, test["review"].size, unlabeled_train["review"].size ))
````

````python
from bs4 import BeautifulSoup
import re

def review_to_wordlist( review):
    review_text = BeautifulSoup(review).get_text()
    review_text = re.sub("[^a-zA-Z]", " ", review_text)
    words = review_text.lower().split()

    return words
````

接下来，我们需要一种特定的输入格式。Word2Vec 期望输入的是单个句子，每个句子作为一个词列表。换句话说，输入格式是一个**列表的列表**。

如何将段落拆分成句子并不是一件简单的事。自然语言中有各种各样的陷阱。英语句子可以以 `?`、`!`、`"`、`.` 等结尾，而且空格和大小写也不是可靠的判断依据。因此，我们将使用 NLTK 的 `punkt` 分词器来进行句子拆分。要使用它，需要安装 NLTK 并使用 `nltk.download()` 下载 `punkt` 的相关训练文件。

````python
import nltk.data

tokenizer = nltk.data.load('tokenizers/punkt/english.pickle')

def review_to_sentences(review, tokenizer):
    #1. Use the NLTK tokenizer to split the paragraph into sentences
    raw_sentences = tokenizer.tokenize(review.strip())

    #2. Loop sentence
    sentences = []
    for raw_sentence in raw_sentences:
        if len(raw_sentence) > 0:
           sentences.append( review_to_wordlist(raw_sentence)) 
            
    return sentences
````

````python
sentences = []

print("Parsing sentences from training set")
for review in train["review"]:
    sentences += review_to_sentences(review, tokenizer)

print("Parsing sentences from unlabeled set")
for review in unlabeled_train["review"]:
    sentences += review_to_sentences(review, tokenizer)
````

## 训练并保存模型

有了经过良好解析的句子列表后，我们就可以开始训练模型了。有许多参数选择会影响运行时间和最终生成模型的质量。
**架构（Architecture）**：架构选项有 skip-gram（默认）或连续词袋（continuous bag of words）。我们发现 skip-gram 稍微慢一点，但产生了更好的结果。

**训练算法（Training algorithm）**：层次 Softmax（hierarchical softmax，默认）或负采样（negative sampling）。对我们来说，默认设置效果很好。

**高频词下采样（Downsampling of frequent words）**：Google 文档推荐的值在 0.00001 到 0.001 之间。对我们来说，接近 0.001 的值似乎能提高最终模型的准确率。

**词向量维度（Word vector dimensionality）**：更多特征会导致更长的运行时间，而且通常（但并非总是）会产生更好的模型。合理的值可以从几十到几百；我们使用了 300。

**上下文/窗口大小（Context / window size）**：训练算法应该考虑多少上下文词？对于层次 Softmax，10 似乎效果很好（在达到某个点之前，越多越好）。

**工作线程数（Worker threads）**：要运行的并行进程数。这因计算机而异，但在大多数系统上 4 到 6 个应该可以工作。

**最低词频（Minimum word count）**：这有助于将词汇表大小限制在有意义的词上。任何在所有文档中出现次数少于这个次数的词都会被忽略。合理的值可以在 10 到 100 之间。在这种情况下，由于每部电影出现 30 次，我们将最低词频设置为 40，以避免对单个电影标题赋予过多重要性。这导致总体词汇量约为 15,000 个词。更高的值也有助于限制运行时间。

## 从词到段落，尝试一：向量平均

IMDB 数据集的一个挑战是评论长度不一。我们需要找到一种方法，将单个词向量转换为每个评论长度相同的特征集。

由于每个词都是 300 维空间中的一个向量，我们可以使用向量运算来组合每条评论中的词。我们尝试的一种方法是将给定评论中的词向量简单地取平均（为此，我们移除了停用词，因为它们只会增加噪音）。

以下代码在第 2 部分代码的基础上，对特征向量进行平均：




````python
import numpy as np

def makeFeatureVec(words,model,num_features):
    featureVec = np.zeros((num_features,),dtype="float32")

    nwords=0

    index2word_set = set(model.wv.index_to_key)


    # 遍历评论中的每个词，如果它在模型的词汇表中，
    # 则将其特征向量加到总和中
    for word in words:
        if word in index2word_set:
            nwords=nwords+1
            featureVec=np.add(featureVec,model.wv[word])

    featureVec = np.divide(featureVec,nwords)


    # 遍历评论中的每个词，如果它在模型的词汇表中，
    # 则将其特征向量加到总和中
    return featureVec


def getAvgFeatureVecs(reviews, model, num_features):
    counter = 0

    reviewFeatureVecs = np.zeros((len(reviews), num_features), dtype="float32")

    for review in reviews:
        if counter%1000 ==0:
            print("Review %d of %d" % (counter, len(reviews)))

        reviewFeatureVecs[counter] = makeFeatureVec(review, model, num_features)

        counter = counter+1

    return reviewFeatureVecs
````

````python
clean_train_reviews = []
for review in train["review"]:
    clean_train_reviews.append(review_to_wordlist(review))

trainDataVecs = getAvgFeatureVecs(clean_train_reviews, model, num_features)

print("Creating average feature vecs for test reviews")

clean_test_reviews = []

for review in test["review"]:
    clean_test_reviews.append( review_to_wordlist(review))

testDataVecs = getAvgFeatureVecs(clean_test_reviews, model, num_features)
````

接下来，使用平均段落向量训练随机森林。请注意，与第 1 部分一样，我们只能使用有标签的训练评论来训练模型：

````python
from sklearn.ensemble import RandomForestClassifier
forest = RandomForestClassifier(n_estimators = 100)

print("Fitting a random forest to labeled training data")
forest = forest.fit( trainDataVecs, train["sentiment"])

result = forest.predict(testDataVecs)

output = pd.DataFrame( data={"id":test["id"], "sentiment":result})
output.to_csv("Word2Vec_AverageVectors.csv", index=False, quoting=3)
````

## 从词到段落，尝试二：聚类
Word2Vec 会创建语义相关的词聚类，因此另一种可能的方法是利用聚类中词语的相似性。这种对向量进行分组的方式被称为“向量量化”。为了实现这一点，我们首先需要找到词聚类的中心，这可以通过使用像 K-Means 这样的聚类算法来完成。

在 K-Means 中，我们需要设置的一个参数是 “K”，即聚类的数量。我们应该如何决定创建多少个聚类呢？反复试验表明，较小的聚类（平均每个聚类只有大约 5 个词）比较大多词聚类能给出更好的结果。

当 K 值很大时，K-Means 聚类可能会非常慢；

````python
from sklearn.cluster import KMeans
import time

start = time.time()

word_vectors = model.wv.vectors
num_clusters = int (word_vectors.shape[0] / 5)

kmeans_clustering = KMeans( n_clusters = num_clusters)
idx = kmeans_clustering.fit_predict(word_vectors)

end = time.time()
elapsed = end - start
print("Time taken for K Means clustering", elapsed, "seconds.")
````

每个词的聚类分配结果现在存储在 `idx` 中，而我们原始 `Word2Vec` 模型中的词汇表仍然存储在 `model.wv.index_to_key `中。为了方便起见，我们将它们打包成一个字典，如下所示：

````python
word_centroid_map = dict(zip(model.wv.index_to_key, idx))
````

这有点抽象，所以让我们仔细看看我们的聚类中包含什么。打印出聚类 0 到 9 的词语：

````python
for cluster in range(0, 10):
    print("\nCluster %d" % cluster)
    # 找出该聚类编号的所有词并打印出来
    words = []
    for i in range(0, len(word_centroid_map.values())):
        if list(word_centroid_map.values())[i] == cluster:
            words.append(list(word_centroid_map.keys())[i])
    print(words)
````

我们可以看到聚类的质量参差不齐。但无论如何，现在每个词都有一个聚类（或“中心”）分配，我们可以定义一个函数，将评论转换为聚类中心袋。这的工作原理与词袋类似，但使用的是语义相关的聚类而不是单个词语：

````python
def create_bag_of_centroids( wordlist, word_centroid_map):
    num_centroids = max( word_centroid_map.values())+1

    bag_of_centroids = np.zeros( num_centroids , dtype="float32")

    for word in wordlist:
        if word in word_centroid_map:
            index = word_centroid_map[word]
            bag_of_centroids[index]+=1

    return bag_of_centroids
````

````python
train_centroids = np.zeros((train["review"].size, num_clusters), dtype="float32")

counter = 0
for review in clean_train_reviews:
    train_centroids[counter] = create_bag_of_centroids(review, word_centroid_map)
    counter += 1

test_centroids = np.zeros((test["review"].size, num_clusters), dtype="float32")

counter = 0
for review in clean_test_reviews:
    test_centroids[counter] = create_bag_of_centroids(review, word_centroid_map)
    counter += 1
````

````python
forest = RandomForestClassifier(n_estimators = 100)

print("Fitting a random forest to labeled training data...")
forest = forest.fit(train_centroids, train["sentiment"])
result =  forest.predict(test_centroids)

output = pd.DataFrame(data={"id":test["id"], "sentiment":result})
output.to_csv("BagOfCentroids.csv", index=False, quoting=3 )
````

## 为什么词袋模型（Bag of Words）表现更好？

最大的原因是，向量平均和聚类中心方法都丢失了词序信息，这使得它们在本质上与词袋模型的概念非常相似。三种方法的效果相近（在标准误差范围内），因此在实际应用中几乎可以视为等价。

可以尝试的几个方向：
第一，在更多文本上训练 Word2Vec 应该能大幅提升性能。Google 的结果是基于在超过十亿词的语料库上学习得到的词向量；而我们使用的有标签和无标签训练集加起来大约只有1800 万词——实在是微不足道。方便的是，Word2Vec 提供了加载任何由 Google 原始 C 工具输出的预训练模型的函数，因此也可以先在 C 中训练模型，然后再导入到 Python 中使用。

第二，在已发表的文献中，分布式词向量技术已被证明优于词袋模型。在这篇论文中，一种名为 Paragraph Vector（段落向量） 的算法在 IMDB 数据集上取得了目前最先进的结果之一。在某种程度上，它比我们这里尝试的方法表现更好，是因为向量平均和聚类丢失了词序，而 Paragraph Vector 保留了词序信息。
